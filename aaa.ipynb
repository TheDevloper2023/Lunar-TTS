```python
# Load Lunar


from api import LunarTTS
from hparams import create_hparams


hparams = create_hparams()
checkpoint_path = "LJ/checkpoint_2000"

tt2 = LunarTTS(checkpoint_path, hparams, "cuda")

```

```python
outputs = tt2("A quick brown fox jumped over the lazy dog", None, True, 0, 2, None, "tpse")
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(8, 6))
plt.imshow(outputs["alignments"].cpu().detach().numpy()[0], aspect='auto', origin='lower', interpolation='none')
plt.xlabel("Encoder timestep (text)")
plt.ylabel("Decoder timestep (mel frames)")
plt.title("Tacotron Alignment")
plt.colorbar()
plt.tight_layout()
plt.show()
```
