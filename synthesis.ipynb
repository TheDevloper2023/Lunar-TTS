```python
# Step 1. Import the necessary libraries
from hparams import create_hparams
from model import load_model
import torch
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
# Step 2. load the nightmare that the uberduck comunity calls "uberduck pipeline v2"

hparams = create_hparams()

model_pth = "LJ/best_val_taco_model" # path to the checkpoint you want to load, can be a relative or absolute path

checkpoint_dict = torch.load(model_pth, map_location='cpu', weights_only=False)

model_dict = checkpoint_dict['state_dict']


model = load_model(hparams)

model.load_state_dict(model_dict)
model.to(device).eval()
```
